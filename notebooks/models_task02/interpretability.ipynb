```python
from utils.interpretability.interpretability import explain_model
from sklearn.exceptions import InconsistentVersionWarning
from utils.data_load.data_load import *
import warnings
warnings.filterwarnings("ignore", category=InconsistentVersionWarning)
```

```python
cur_dir = Path(os.getcwd()).resolve()
proj_dir = next(parent for parent in [cur_dir, *cur_dir.parents] if (parent / "utils").exists() and (parent / "data").exists())
SELECTED_SUBJ = "chb12"
SPLIT_PATH = proj_dir / "data" / "split" / "task_2" 
MODEL_PATH = proj_dir / "saved_models" / "task_2"
SUBJ_PATH = SPLIT_PATH / SELECTED_SUBJ
SELECTED_FILES = sorted([f.stem for f in SUBJ_PATH.glob("*.npz")]) # ["chb04_08", "chb04_05", "chb04_01", "chb04_02"]
```

# Load Data

```python
# 1. Load and Pool
loaded = [load_file_pooled(f_name, path=SUBJ_PATH) for f_name in SELECTED_FILES]
# 2. Global Concatenation
split_frames = {
    split: pd.concat([f[split] for f in loaded], ignore_index=True)
    for split in ["train", "val", "test"]
}
```

# Load Models

```python
models = load_models(MODEL_PATH)
```

```python
train = split_frames["train"]
test = split_frames["test"]
```

# Interpretation of the model

```python
for model in models:
    X_train, _ = model.preprocess(train, is_training=False)
    X_test, y_test = model.preprocess(test, is_training=False)
    try:
        feature_names = model.features_selected
    except Exception as e:
        feature_names = test.columns[:-1]
    
    X_train = pd.DataFrame(X_train, columns=feature_names)
    X_test = pd.DataFrame(X_test, columns=feature_names)

    explain_model(X_train=X_train, X_test=X_test, y_test=y_test, model=model)
```
